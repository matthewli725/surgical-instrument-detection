# Robustness of Object Detection Models to Specular Lighting on Metallic Objects

## Problem Statement

Modern computer vision models often assume relatively stable appearance conditions. However, **metallic objects**—such as surgical instruments—exhibit strong **specular reflections**, making their appearance highly sensitive to lighting direction, intensity, and environment.

This introduces a critical failure mode:
- highlights can saturate pixels  
- reflections can obscure shape cues  
- appearance can vary dramatically across lighting conditions  

As a result, object detection and classification models may degrade significantly under adverse lighting.

We aim to evaluate the following hypothesis:

> Object detection models that are robust to specular lighting variation on stainless-steel objects will generalize to similar metallic objects, such as surgical instruments.

---

## Key Insight

Rather than explicitly modeling material reflectance (e.g., BRDF), we focus on **image-space manifestations of specularity**:
- highlight intensity and saturation  
- spatial extent of reflections  
- contrast distortions  

These are the actual factors that affect model performance.

---

## Proposed Approach

We design a **controlled, lightweight experimental pipeline** to measure model robustness under varying lighting conditions using accessible proxy objects (kitchen silverware).

---

## Experimental Design

### 1. Data Collection

We collect a small dataset of stainless-steel objects (e.g., spoons, forks, knives) under controlled lighting variation.

**Setup:**
- Fixed camera position  
- Static background  
- 2–5 object instances  
- 5–10 distinct lighting configurations  

**Lighting variations:**
- Direction (azimuth / elevation)  
- Intensity (low → saturation)  
- Type (point vs diffuse)  
- Angle (including grazing angles for strong highlights)  

**Dataset size:**
- ~50–150 images total  

Bounding boxes are labeled for each object.

---

### 2. Model Selection

We use a **pretrained object detector** (e.g., YOLO, Detectron2).

No training or fine-tuning is required for the baseline experiment.

---

### 3. Evaluation Protocol

For each lighting condition:
- Run inference on all images  
- Record:
  - detection success (precision/recall or AP if available)  
  - confidence scores  
  - failure cases  

We analyze how performance changes as lighting becomes more challenging.

---

### 4. Specularity Quantification

To link performance degradation to reflectance effects, we compute simple **image-based specularity metrics**:

- Percentage of saturated pixels within object bounding boxes  
- Highlight area fraction (bright regions above threshold)  
- Intensity statistics (e.g., top percentile brightness)  

We then analyze:
- model performance vs specularity level  

---

### 5. Optional External Validation

To strengthen the claim, we optionally evaluate on:
- a small set of surgical instrument images (e.g., collected online)

We compare:
- qualitative detection behavior  
- specularity statistics relative to the silverware dataset  

---

## Expected Outcomes

This experiment allows us to:

- Quantify how detection performance changes under specular lighting  
- Identify whether specularity is a dominant failure mode  
- Demonstrate robustness (or lack thereof) to metallic reflectance effects  

If the model maintains performance across strong specular conditions, this supports the hypothesis that it will generalize to similar metallic objects.

---

## Scope and Limitations

- We do **not** attempt to recover physical material properties (e.g., BRDF)  
- Results are **condition-dependent** and based on controlled lighting variation  
- Silverware serves as a **proxy**, not a perfect substitute for surgical instruments  

---

## Summary

This project provides a **practical and focused evaluation** of model robustness to specular lighting effects using:
- minimal data collection  
- controlled experimental conditions  
- direct measurement of performance degradation  

The goal is to bridge the gap between theoretical reflectance properties and real-world model behavior in challenging visual conditions.
